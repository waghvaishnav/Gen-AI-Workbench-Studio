# Temperature :

```python
from transformers import pipeline


pipe = pipeline("text-generation",model="gpt2")

pipe("here is a suggestion on my new coffee shop name, the name should be:",temperature=1.5,max_new_tokens=10)
```

# TOP -P  :

```python
pipe = pipeline("text-generation",model="gpt2")

pipe("here is a suggestion on my new coffee shop name, the name should be:",max_new_tokens=10,top_p=0.6)
```

# TOP_K  :

```python


pipe = pipeline("text-generation",model="gpt2")

pipe("The cat sat on the",max_new_tokens=10,top_k=1)
```

## When we use the top_k = 1 , it always resulted in same output prediction every Time.

```python


pipe = pipeline("text-generation",model="gpt2")

pipe("The cat sat on the",max_new_tokens=10,top_k=5)
```

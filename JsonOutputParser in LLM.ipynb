```python
from dotenv import load_dotenv

load_dotenv()
```

```python
# importing required modules :

from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import JsonOutputParser
from langchain_core.exceptions import OutputParserException

# import and load groq model here :
from langchain_groq import ChatGroq


llm = ChatGroq(model="openai/gpt-oss-120b")

```

```python
# creating a Prompt Template : 

prompt = """ from the following data article extracts the information such as "revenue expected", "revenue actual", "eps actual" , "eps expected",
          the extracted info should be in json format 
           article is :{article}      """


pr_tem = PromptTemplate.from_template(prompt)

pr_tem
```

```python
# create a chain here :

chain = pr_tem | llm

chain
```

```python
article = """  EPS estimated vs expected and revenue expected vs actual
Tesla's Q2 2026 earnings report revealed a mixed performance, with revenue exceeding estimates but earnings falling
short. The company reported a revenue of $28.24 billion, up 25.5% year-over-year, which was ahead of the $26.36 billion consensus
estimate by 7.10%. However, the non-GAAP EPS of $0.33 missed the $0.54 estimate by 38.51%. This discrepancy was attributed to a 47%
increase in operating expenses, driven by AI infrastructure buildout, R&D investment, and stock-based compensation tied to the 2025
CEO Performance Award. The aggressive spending pressured operating margins to just 1.4%, despite record deliveries of 480,126 vehicles
and 13.5 GWh of energy storage deployments. Free cash flow turned negative at $1.09 billion, reversing a previous positive trend. Looking
ahead, Tesla expects hardware profits to be gradually accompanied by accelerating AI, software, and fleet-based revenues as its Robotaxi
network expands and Optimus production ramps through 2026.  """

total_estimate = chain.invoke({"article":article})

print(total_estimate.content)
```

```python
type(total_estimate.content)
```

### in the above example we give prompt for give output of the above prompt in a "json" format. it give json but inside the string.

## To Overcome this problem we have a concept called "JsonOutputParser" that converts the model output into "Json" format.

```python
# parser = JsonOutputParser()

try:

    parser = JsonOutputParser()
    
    json_format_estimate = parser.parse(total_estimate.content)
    
    print(json_format_estimate)
except Exception as e:
    raise OutputParserException(f"output parser not working properly.{e}")
```

### Here by using JsonOutputParser we can Successfully convert output into "Json".

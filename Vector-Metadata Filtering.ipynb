# Metadata Filtering (Where Clause ):

```python
from datetime import datetime

client = chromadb.Client()

collection = client.create_collection("reviews",
                                     metadata={
                                         "description":"review data",
                                         "created_on":str(datetime.now())
                                     })
```

```python
collection.add(documents=[
    "The wireless headphones have excellent sound quality and comfortable ear cushions.",
    "The laptop battery lasts for a long time, but the keyboard could be better.",
    "The running shoes are lightweight and comfortable, making them great for daily workouts.",
    "The coffee maker is easy to use, but it takes a little longer to brew coffee."
],ids = ["id1","id2","id3","id4"],
              metadatas=[{"category":"electronics","rating":4},
                         {"category":"electronics","rating":3},
                         {"category":"sports","rating":4},
                         {"category":"home_appliences","rating":2}])

```

```python
collection.query(
    query_texts="",
    where={"rating":{"$in":[1,6,2]}}
    
)
```

```python
collection.query(
    query_texts="machines for driving car",
    where={"rating":{"$lt":3}}
    
)
```

## operators for where clause :

$in : in

$gt : greater than

$lt : less than

$nin : not in 

# Filtering using Where Documents :

```python
collection.query(
    query_texts = "my best fruit is apple",
    where_document={"$contains":"coffee"}
)
```

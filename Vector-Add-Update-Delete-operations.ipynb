# Add/ Update / Delete in Chromadb :

```python
import chromadb
```

```python
client = chromadb.Client()

from datetime import datetime

client.create_collection(name="news_data",
                        metadata={
                            "description":"product reviews news data",
                            "created":str(datetime.now())
                        })
```

```python
collection = client.get_collection("news_data")

collection
```

```python
client.list_collections()
```

# Add :

```python
collection.add(documents=[
    "The wireless headphones have excellent sound quality and comfortable ear cushions.",
    "The laptop battery lasts for a long time, but the keyboard could be better.",
    "The running shoes are lightweight and comfortable, making them great for daily workouts.",
    "The coffee maker is easy to use, but it takes a little longer to brew coffee."
],ids = ["id1","id2","id3","id4"],
              metadatas=[{"category":"electronics","rating":4},
                         {"category":"electronics","rating":3},
                         {"category":"sports","rating":4},
                         {"category":"home_appliences","rating":2}])

```

```python
collection.query(
    query_texts="the sound quality is very awesome!",
    n_results=2
)
```

```python
collection.query(
    query_texts="the gpu configuration is very high",
    n_results=1
)
```

# Update :

```python
collection.update(documents=["The wireless headphones have awesome sound quality and comfortable ear cushions."],
                 ids=["id1"],
                 metadatas=[{"category":"electronics","rating":5}])
```

```python
collection.get(ids=["id1"])
```

```python
collection.query(
    query_texts="data science",
    n_results=5,
    where={"category":"sports"}
)
```

# deleting specific ids documents :

```python
collection.delete(ids=["id1"])
```

# deleting entire collection :

```python
client.delete_collection("news_data")
```

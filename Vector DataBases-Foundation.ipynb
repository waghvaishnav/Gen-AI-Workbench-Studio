```python
import chromadb
```

```python
client = chromadb.Client()

client.create_collection(name="news")
```

```python
# for delete collection :

client.delete_collection("news")
```

```python
collection = client.get_collection("news")      # two ways of define collecttion,direct or get_collection.

collection
```

```python
# for adding the documents in the collections.

collection.add(
    documents=[
        "machine learning is the hottest career in 2026",
        "india is the developing country in world.",
        "the farmer is produces crops for the food."
    ],
    ids = ["id1","id2","id3"]
)
```

```python
# for query searching for similarity in documents :

collection.query(
    query_texts=[
        "computer science"
    ],
    n_results=2
)
```

```python
collection.query(
    query_texts=[
        "water bottles"
    ],
    n_results=3
)
```

```python
collection.delete(ids=["id3"])
```

```python
collection.count()
```

```python
client.heartbeat()
```

# PersistentClient :

```python
clientp = chromadb.PersistentClient(path="./vector_data")
```

```python
from dotenv import load_dotenv

load_dotenv()
```

```python
# import embeddings :
from chromadb.utils import embedding_functions

ef = embedding_functions.SentenceTransformerEmbeddingFunction(model_name="all-distilroberta-v1")


collection = clientp.create_collection(name = "new_vector",
                                      embedding_function=ef)   # provides embedding function to the client created.
```

```python
collection.add(documents=[
    "communication is the most important skills",
    "python for data science",
],ids=["id1","id2"])

```

```python
clientp.heartbeat()
```

```python
clientp.reset        # reset all or deleting all collection/documents in client.
```

```python
clientp.list_collections()
```

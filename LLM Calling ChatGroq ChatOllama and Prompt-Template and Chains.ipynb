# Prompt Templates and Chains :

```python
from langchain_core.prompts import PromptTemplate

pt = PromptTemplate.from_template("suggest a clinical tablets on {topic} in 40 words")
```

```python
llm = ChatGroq(model="openai/gpt-oss-120b")

llm
```

```python
chain = pt | llm

response = chain.invoke({"topic":"high fever"})
```

```python
response
```

```python
print(response.content)
```

```python
response1 = chain.invoke({"topic":"typhoid"})
```

```python
response1.content
```

```python
pt1 = PromptTemplate.from_template("write a poem for love between {topic1} and {topic2}")

pt1
```

```python
poem = pt1 | llm

result = poem.invoke({"topic1":"laptop",
     "topic2":"mouse_laptop"})

print(result.content)
```

```python
result2 = poem.invoke({"topic1":"farmer",
              "topic2":"crops"})

print(result2.content)
```
